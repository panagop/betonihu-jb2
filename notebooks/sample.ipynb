```python
import math

# Calculate the area of a circle
radius = 5
area = math.pi * radius ** 2
print(f"The area of a circle with radius {radius} is {area:.2f}")
```

## A Simple Plot

Let's create a basic plot using `matplotlib`.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.set_title("Sine Wave")
plt.show()
```

## A Markdown Table

|----------|:--------------:|:------------:|

| C25/30   | 25             | 31           |
| C30/37   | 30             | 33           |

```python
# Simple calculation
fck = 30  # MPa
fcm = fck + 8  # Mean compressive strength
Ecm = 22 * (fcm / 10) ** 0.3  # GPa

print(f"Concrete class: C{fck}/{int(fck * 1.25)}")
print(f"Mean compressive strength fcm = {fcm} MPa")
print(f"Modulus of elasticity Ecm = {Ecm:.2f} GPa")
```
