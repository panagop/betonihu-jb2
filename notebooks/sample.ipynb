## A Simple Plot

Let's create a basic plot using `matplotlib`.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.set_title("Sine Wave")
plt.show()
```

## Using the `betonihu` Package

The `betonihu` package (from `src/betonihu/`) provides reusable classes for concrete and steel material properties per Eurocode 2.

```python
from betonihu import ConcreteProperties, SteelProperties

# Define materials
concrete = ConcreteProperties(fck=30)
steel = SteelProperties(fyk=500)

print(f"Concrete: {concrete}")
print(f"  Mean tensile strength fctm = {concrete.fctm:.2f} MPa")
print()
print(f"Steel: {steel}")
print(f"  Design yield strain εyd = {steel.eyd:.5f}")
```

```python
import matplotlib.pyplot as plt

# Compare concrete classes
classes = [20, 25, 30, 35, 40, 45, 50]
props = [ConcreteProperties(fck=f) for f in classes]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.bar([f"C{p.fck:.0f}" for p in props], [p.fcd for p in props], color="steelblue")
ax1.set_ylabel("fcd [MPa]")
ax1.set_title("Design Compressive Strength")

ax2.bar([f"C{p.fck:.0f}" for p in props], [p.Ecm for p in props], color="coral")
ax2.set_ylabel("Ecm [GPa]")
ax2.set_title("Modulus of Elasticity")

plt.tight_layout()
plt.show()
```

```python
# Simple calculation
fck = 30  # MPa
fcm = fck + 8  # Mean compressive strength
Ecm = 22 * (fcm / 10) ** 0.3  # GPa

print(f"Concrete class: C{fck}/{int(fck * 1.25)}")
print(f"Mean compressive strength fcm = {fcm} MPa")
print(f"Modulus of elasticity Ecm = {Ecm:.2f} GPa")
```
